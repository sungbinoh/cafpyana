```python
import matplotlib.pyplot as plt
import numpy as np
import mpltern

# 1. Apply maximum threshold cuts to clean raw outliers
cut_df = clean_trkdf[
    (clean_trkdf.chi2p <= 500)
    & (clean_trkdf.chi2u <= 100)
    & (clean_trkdf.chi2pi <= 100)
].copy()

# 4. Filter by true particle types
clean_trkdf_true_mu = cut_df[cut_df.true_pdg == 13]
clean_trkdf_true_p = cut_df[cut_df.true_pdg == 2212]
clean_trkdf_true_pi = cut_df[
    (cut_df.true_pdg == 211) | (cut_df.true_pdg == -211)
]

ax = plt.subplot(projection="ternary")

# 5. Plot equal-weight scaled variables within the cut limits
ax.scatter(
    clean_trkdf_true_mu.chi2p,
    clean_trkdf_true_mu.chi2u,
    clean_trkdf_true_mu.chi2pi,
    alpha=0.4,
    label="True Muons",
)
ax.scatter(
    clean_trkdf_true_p.chi2p,
    clean_trkdf_true_p.chi2u,
    clean_trkdf_true_p.chi2pi,
    alpha=0.4,
    label="True Protons",
)
ax.scatter(
    clean_trkdf_true_pi.chi2p,
    clean_trkdf_true_pi.chi2u,
    clean_trkdf_true_pi.chi2pi,
    alpha=0.4,
    label="True Pions",
)

ax.legend(bbox_to_anchor=(0.8, 0.9), loc="upper left")
ax.set_tlabel(r"$\chi^2_p$")
ax.set_llabel(r"$\chi^2_\mu$")
ax.set_rlabel(r"$\chi^2_\pi$")
ax.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
training_cols = ["chi2u", "chi2p"]
truth_cols = ["true_pdg"]
selected_cols = list(set(training_cols + truth_cols))

clean_trkdf = trkdf[(trkdf.true_pdg == -211) | (trkdf.true_pdg == 13)][selected_cols]
le = LabelEncoder()
clean_trkdf["target"] = le.fit_transform(clean_trkdf["true_pdg"])

full_reco_df = clean_trkdf[training_cols]
full_truth_df = clean_trkdf["target"]

train_df, test_df, train_labels, test_labels = train_test_split(full_reco_df, full_truth_df, test_size=.5, random_state=123)
```

```python
# 1. Create GPU-optimized QuantileDMatrix objects
sample_weights = compute_sample_weight("balanced", train_labels)
dtrain = xgb.QuantileDMatrix(train_df, label=train_labels, weight=sample_weights)
dtest = xgb.QuantileDMatrix(test_df, label=test_labels)

# 2. Configure GPU training parameters
params = {
    "device": "cuda",
    "tree_method": "hist",
    "objective": "multi:softprob",  # Adjust objective based on your target setup (e.g., binary:logistic or multi:softprob)
    "num_class": 2
}

# 3. Train the XGBoost model
booster = xgb.train(params, dtrain, num_boost_round=100)

# 4. Generate predictions
preds = booster.predict(dtest)
```

```python
pdg_map = dict(zip(range(len(le.classes_)), le.classes_))
plt.title("BDT Score, Excluding MCS")
plt.hist(preds, stacked=True, label=pdg_map.values())
plt.legend()
```
